```python
import pandas as pd

df = pd.read_csv("SuperMarket Analysis.csv")

df.head()
```

```python
# Check missing values
missing_values = df.isnull().sum()

print(missing_values)
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())
```

```python
# Check current data types
print(df.dtypes)
```

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'])

# Convert Time column to datetime
df['Time'] = pd.to_datetime(df['Time']).dt.time

# Check the result
print(df[['Date', 'Time']].head())
```

```python
# Convert Date column to datetime
df['Date'] = pd.to_datetime(df['Date'], format='%Y-%m-%d')

# Convert Time column to time
df['Time'] = pd.to_datetime(df['Time'], format='%H:%M:%S').dt.time

# Check the result
print(df[['Date', 'Time']].head())
```

Now that the 'Date' and 'Time' columns are in the correct format, we can extract additional time-based features that might be useful for analysis or ELT processes.

```python
# Extract hour from the 'Time' column
df['Hour'] = df['Time'].apply(lambda x: x.hour)

# Extract day of the week (0=Monday, 6=Sunday) from the 'Date' column
df['DayOfWeek'] = df['Date'].dt.dayofweek

# Extract day of the month from the 'Date' column
df['Day'] = df['Date'].dt.day

# Extract month from the 'Date' column
df['Month'] = df['Date'].dt.month

# Extract year from the 'Date' column
df['Year'] = df['Date'].dt.year

# Display the new columns along with 'Date' and 'Time' to verify
print(df[['Date', 'Time', 'Hour', 'DayOfWeek', 'Day', 'Month', 'Year']].head())
```

```python
from sklearn.preprocessing import OneHotEncoder

# Identify categorical columns for one-hot encoding
categorical_cols = ['Branch', 'City', 'Customer type', 'Gender', 'Product line', 'Payment']

# Create a OneHotEncoder object
ohe = OneHotEncoder(handle_unknown='ignore', sparse_output=False)

# Fit and transform the categorical columns
encoded_features = ohe.fit_transform(df[categorical_cols])

# Create a DataFrame from the encoded features
encoded_df = pd.DataFrame(encoded_features, columns=ohe.get_feature_names_out(categorical_cols))

# Concatenate the encoded DataFrame with the original DataFrame, dropping original categorical columns
df_transformed = pd.concat([df.drop(columns=categorical_cols), encoded_df], axis=1)

# Display the first few rows of the transformed DataFrame and its info to see the new columns and types
print("Transformed DataFrame Head:")
print(df_transformed.head())
print("\nTransformed DataFrame Info:")
df_transformed.info()
```

### **Transform: Data Aggregation**


```python
# Aggregate daily sales by Branch and Product Line
df_agg_sales = df_transformed.groupby(['Date', 'Branch_Alex', 'Branch_Cairo', 'Branch_Giza', 'Product line_Electronic accessories', 'Product line_Fashion accessories', 'Product line_Food and beverages', 'Product line_Health and beauty', 'Product line_Home and lifestyle', 'Product line_Sports and travel']).agg(
    TotalSales=('Sales', 'sum'),
    AverageRating=('Rating', 'mean'),
    TotalQuantity=('Quantity', 'sum')
).reset_index()

print("Aggregated Daily Sales by Branch and Product Line:")
print(df_agg_sales.head())

df_agg_sales.info()
```

### **Load: Save Transformed Data**



```python
# Define the output file name
output_filename = 'transformed_supermarket_sales.csv'

# Save the transformed (and aggregated) DataFrame to a new CSV file
df_agg_sales.to_csv(output_filename, index=False)

print(f"Transformed data successfully loaded to '{output_filename}'")
```

```python
print("Starting data validation...")

# Define validation rules
# Rule 1: 'Sales' must be positive
# Rule 2: 'Quantity' must be a non-negative integer (already int64, so check >= 0)
# Rule 3: 'Rating' must be between 1 and 10 (inclusive)

# Apply validation rules
valid_sales = df_transformed['Sales'] > 0
valid_quantity = df_transformed['Quantity'] >= 0
valid_rating = (df_transformed['Rating'] >= 1) & (df_transformed['Rating'] <= 10)

# Combine all validation rules
all_valid_rules = valid_sales & valid_quantity & valid_rating

# Separate valid and invalid rows
df_valid = df_transformed[all_valid_rules].copy()
df_invalid = df_transformed[~all_valid_rules].copy()

print(f"Total records: {len(df_transformed)}")
print(f"Valid records: {len(df_valid)}")
print(f"Invalid records: {len(df_invalid)}")

if not df_invalid.empty:
    print("\nInvalid records found. Displaying first 5 invalid records:")
    print(df_invalid.head())
else:
    print("\nNo invalid records found. All data passed validation.")
```

```python
# Create date-related attributes

df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Month_Name'] = df['Date'].dt.month_name()
df['Quarter'] = df['Date'].dt.quarter
df['Day'] = df['Date'].dt.day
df['Day_Name'] = df['Date'].dt.day_name()
df['Day_Of_Week'] = df['Date'].dt.dayofweek + 1

print(df[
    [
        'Date',
        'Year',
        'Month',
        'Month_Name',
        'Quarter',
        'Day',
        'Day_Name',
        'Day_Of_Week'
    ]
].head())
```

```python
# Extract hour
df['Hour'] = df['Time'].apply(lambda x: x.hour)

# Create time period
def get_time_period(hour):
    if hour < 12:
        return 'Morning'
    elif hour < 17:
        return 'Afternoon'
    elif hour < 21:
        return 'Evening'
    else:
        return 'Night'

df['Time_Period'] = df['Hour'].apply(get_time_period)

print(df[['Time', 'Hour', 'Time_Period']].head())
```

```python
# Calculate total cost before tax
df['Calculated_COGS'] = df['Unit price'] * df['Quantity']

# Calculate tax from COGS
df['Calculated_Tax'] = df['Calculated_COGS'] * 0.05

# Calculate gross profit
df['Calculated_Gross_Profit'] = df['Sales'] - df['Calculated_COGS']

print(
    df[
        [
            'Unit price',
            'Quantity',
            'Sales',
            'Calculated_COGS',
            'Calculated_Tax',
            'Calculated_Gross_Profit'
        ]
    ].head()
)
```

```python
print("Starting data validation...")

# Sales must be greater than 0
valid_sales = df['Sales'] > 0

# Quantity must be greater than 0
valid_quantity = df['Quantity'] > 0

# Rating must be between 1 and 10
valid_rating = df['Rating'].between(1, 10)

# Unit price must be greater than 0
valid_price = df['Unit price'] > 0

# Combine validation rules
all_valid = (
    valid_sales &
    valid_quantity &
    valid_rating &
    valid_price
)

df_valid = df[all_valid].copy()
df_invalid = df[~all_valid].copy()

print("Total records :", len(df))
print("Valid records :", len(df_valid))
print("Invalid records:", len(df_invalid))
```

```python
if len(df_invalid) == 0:
    print("✅ All records passed validation.")
else:
    print("⚠️ Invalid records found.")
    display(df_invalid.head())
```
